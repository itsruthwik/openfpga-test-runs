```python
# --- Configuration ---

# Define the list of hard logic area values to check
list_of_hard_logic_areas = [29000, 46000, 40000, 57000]

# Define the maximum values for M and N to search up to
maximum_M_to_check = 20
maximum_N_to_check = 20

# --- Finding optimal M,N values ---

print("Finding rectangular configurations (M,N) close to squares...")
print("-" * 70)

for current_hard_logic_area in list_of_hard_logic_areas:
    print(f"Checking for hard logic area = {current_hard_logic_area}:")
    
    best_M = None
    best_N = None
    best_score = float('inf')
    
    # Iterate through possible values for N and M (where M ≠ N)
    for N in range(1, maximum_N_to_check + 1):
        for M in range(1, maximum_M_to_check + 1):
            if M == N:  # Skip square configurations
                continue
                
            # Calculate left side: 2239*N*N (based on the smaller dimension)
            left_side = 2239 * min(M, N) * min(M, N)
            
            # Calculate right side: hard logic area + (M+N)*(303) + (M*N)*(688)
            right_side = current_hard_logic_area + (M+N)*303 + (M*N)*688
            
            # Calculate the absolute difference between expressions
            expression_diff = abs(left_side - right_side)
            
            # Calculate how far the rectangle is from being a square
            # The closer to 1, the more square-like the rectangle is
            square_ratio = max(M, N) / min(M, N)
            squareness_penalty = (square_ratio - 1) * 1000  # Scale factor to balance with expression_diff
            
            # Combined score (lower is better)
            combined_score = expression_diff + squareness_penalty
            
            # Check if this is the best configuration so far
            if combined_score < best_score:
                best_score = combined_score
                best_M = M
                best_N = N
    
    # Print the results
    print(f"  Best rectangular configuration: M = {best_M}, N = {best_N}")
    print(f"  Rectangle aspect ratio: {max(best_M, best_N) / min(best_M, best_N):.2f}")
    left_side = 2239 * min(best_M, best_N) * min(best_M, best_N)
    right_side = current_hard_logic_area + (best_M+best_N)*303 + (best_M*best_N)*688
    print(f"  Left side (2239*min(M,N)²): {left_side}")
    print(f"  Right side (area + 303*(M+N) + 688*M*N): {right_side}")
    print(f"  Expression difference: {abs(left_side - right_side)}")
    print("-" * 70)

print("Analysis complete.")
```

```python
# --- Configuration ---

# Define the list of hard logic area values to check
list_of_hard_logic_areas = [29000, 46000, 40000, 57000]

# Define the maximum values for M and N to search up to
maximum_M_to_check = 20
maximum_N_to_check = 20

# --- Finding optimal M,N values ---

print("Finding square-like configurations (M,N) where 2239*min(M,N)² > (area + other terms)...")
print("-" * 70)

for current_hard_logic_area in list_of_hard_logic_areas:
    print(f"Checking for hard logic area = {current_hard_logic_area}:")
    
    best_M = None
    best_N = None
    best_score = float('inf')
    found_any = False
    
    # Iterate through possible values for N and M
    for N in range(1, maximum_N_to_check + 1):
        for M in range(1, maximum_M_to_check + 1):
            # Calculate left side: 2239*min(M,N)²
            left_side = 2239 * min(M, N) * min(M, N)
            
            # Calculate right side: hard logic area + (M+N)*(303) + (M*N)*(688)
            right_side = current_hard_logic_area + (M+N)*303 + (M*N)*688
            
            # Only consider configurations where left side > right side
            if left_side > right_side:
                found_any = True
                
                # Calculate how close the configuration is to being a square
                square_ratio = max(M, N) / min(M, N)
                
                # Strong preference for square-like configurations
                # Using squareness_penalty^2 to heavily penalize non-square configurations
                squareness_penalty = ((square_ratio - 1) * 2000) ** 2
                
                # Calculate margin (how much left_side exceeds right_side)
                margin = left_side - right_side
                
                # Prefer smaller margins (closer to threshold) while maintaining square-like configs
                combined_score = margin + squareness_penalty
                
                # Check if this is the best configuration so far
                if combined_score < best_score:
                    best_score = combined_score
                    best_M = M
                    best_N = N
    
    # Print the results
    if found_any:
        print(f"  Best configuration: M = {best_M}, N = {best_N}")
        print(f"  Aspect ratio: {max(best_M, best_N) / min(best_M, best_N):.2f}")
        left_side = 2239 * min(best_M, best_N) * min(best_M, best_N)
        right_side = current_hard_logic_area + (best_M+best_N)*303 + (best_M*best_N)*688
        print(f"  Left side (2239*min(M,N)²): {left_side}")
        print(f"  Right side (area + 303*(M+N) + 688*M*N): {right_side}")
        print(f"  Margin (left - right): {left_side - right_side}")
    else:
        print("  No configurations found where left side > right side within the search range.")
    
    print("-" * 70)

print("Analysis complete.")
```
